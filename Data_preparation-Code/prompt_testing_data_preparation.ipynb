```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
import pandas as pd

# Assuming model and tokenizer are already initialized
# model = ...
# tokenizer = ...

# Initialize an empty list to store generated steps
generated_steps = []

# Loop over 10 validation cases (you can modify this as needed)
for i in range(194):
    # Modify your input text for each validation case
    input_text = f"""
    <|begin_of_text|><|start_header_id|>system<|end_header_id|>
    Generate steps which need to be given in a Llama3.1 model to generate the exact same code. Steps need to be compact and as minimal as possible to generate the exact same code.
    Format of the output should be:
    Name: Name of the file (Example: LXOXF-xxx or ULSFQA-xxx)
    Test Case: Description of the test case in one line.
    Step 1: Description of the first step.
    Step 2: Description of the second step.
    In this way, steps will continue.  
    {train['Python Code'][i]}
    The number of steps should be as minimal as possible, but it should be sufficient. In the output, only give Name, Test Case, and Steps. I only need steps. No code should be there in your output.<|eot_id|>
    <|start_header_id|>user<|end_header_id|>
    """

    # Tokenize input
    input_ids = tokenizer.encode(input_text, return_tensors="pt")

    # Generate output (text completion or response)
    output_ids = model.generate(input_ids, max_length=16384, num_return_sequences=1, pad_token_id=tokenizer.eos_token_id)

    # Decode only the generated part (excluding the input)
    output_text = tokenizer.decode(output_ids[0][len(input_ids[0]):], skip_special_tokens=True)

    # Append the generated steps to the list
    generated_steps.append(output_text)

    print(f"Generated steps for validation case {i+1} added.")

# Create a DataFrame to hold the generated steps
df = pd.DataFrame(generated_steps, columns=['Generated_steps'])

# Save the DataFrame to an Excel file
df.to_excel('G_train_Steps_1.xlsx', index=False)

print("Excel file 'G_train_Steps_1.xlsx' created successfully with generated steps.")
```
